```python
from data_handler.data_loader import load_monk
from nn.model import Model
from nn.layers import Dense, xavier_uniform
from nn.activations import Tanh, Sigmoid
from nn.dropout import Dropout
from nn.losses import BinaryCrossEntropy
from nn.optim import SGD
from nn.metrics import Accuracy
from nn.callbacks import EarlyStopping

from training.trainer import Trainer

# Load data
X_train, y_train = load_monk("data/MONK/MONK1/monks-1.train")
X_test,  y_test  = load_monk("data/MONK/MONK1/monks-1.test")

# Build model
model = Model(
    modules=[
        Dense(17, 16, initializer=xavier_uniform),
        Tanh(),
        Dropout(p=0.2, seed=123),
        Dense(16, 1, initializer=xavier_uniform),
        Sigmoid(),
    ],
    loss=BinaryCrossEntropy(),
    optimizer=SGD(lr=0.1),
    metrics=[Accuracy()],
    callbacks=[EarlyStopping(monitor="val_loss", patience=20, verbose=1)],
)

# Train
trainer = Trainer(model, verbose=1)
history = trainer.fit(
    X_train, y_train,
    X_val=X_test, y_val=y_test,   # (temporary; later make a proper val split)
    epochs=200,
    batch_size=32,
    shuffle=True,
    seed=42,
)

# Evaluate
print(trainer.evaluate(X_test, y_test))
```
