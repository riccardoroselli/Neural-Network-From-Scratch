```python
import os
from experiments.MONK.common import run_monk_selection

SEEDS = [42]
TOP_K = 10
```

```python
CONFIG_MONK1 = "experiments/MONK/configs/monk1.yaml"
OUT_MONK1 = "experiments/MONK/results/monk1"

os.makedirs(OUT_MONK1, exist_ok=True)

print("\n==============================")
print("Running MONK-1 model selection")
print("==============================")

out_monk1 = run_monk_selection(
    config_path=CONFIG_MONK1,
    out_dir=OUT_MONK1,
    seeds=SEEDS,
    top_k=TOP_K,
    n_jobs=-1
)

print("MONK-1 DONE")
print("Best score:", out_monk1["best_score"])
```
